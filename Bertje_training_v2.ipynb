```python
# Train SBERT-style classifier on EITHER cosine-labeled OR regex-labeled datasets (toggle via CONFIG)
# Robust path resolution + checkpoint auto-creation (tokenizer+config) + safe loading
# Loss entirely in CustomLossTrainer. Includes "Unassigned" gate + per-class threshold calibration.
# Safe best-model selection via eval_loss (guaranteed), with optional switch to F1 after verifying metrics.

import os, json, ast, glob
import numpy as np
import pandas as pd
import torch
from torch import nn
from datasets import Dataset
from transformers import (
    AutoModel,
    AutoTokenizer,
    AutoConfig,
    TrainingArguments,
    Trainer,
    EvalPrediction
)
from transformers.modeling_outputs import SequenceClassifierOutput
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt
import seaborn as sns
# 1) ADD this import (near your other HF imports)
from transformers import DataCollatorWithPadding

from transformers.trainer_callback import PrinterCallback

# =========================
# CONFIG
# =========================
LABELING_SYSTEM = "regex"    # "cosine", "regex", or "auto"
USE_MULTI_LABEL = "true"     # "auto" | True | False

# SBERT backbone (Dutch-capable encoders recommended)
SBERT_MODEL_NAME = "GroNLP/bert-base-dutch-cased"
# Alternatives: "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2", etc.

CHECKPOINT_BASE    = r"C:\Users\Home\policy-analysis\Modeling\Regex_pretraining_slavery_sentences_v4\bertje_checkpoints"
CHECKPOINT_SAVE_TO = r"C:\Users\Home\policy-analysis\Modeling\Regex_pretraining_slavery_sentences_v4\bertje_checkpoints"

# Cosine defaults
COSINE_DIR        = r"C:\Users\Home\policy-analysis\Corex_topicdata_slavery_v5"
COSINE_TRAIN_PATH = os.path.join(COSINE_DIR, "train_BERTJE_CHUNKS.csv")
COSINE_VAL_PATH   = os.path.join(COSINE_DIR, "val_BERTJE_CHUNKS.csv")
COSINE_META_PATH  = r"C:\Users\Home\policy-analysis\Corex_topicdata_Slavery\bertje_split_meta.json"

# Regex preferred dir
PREFERRED_REGEX_DIR = r"C:\Users\Home\policy-analysis\Modeling\Regex_pretraining_slavery_sentences_4\Bertje_prep"

# "Unassigned" (NONE-OF-THE-ABOVE) gate
USE_UNASSIGNED_GATE         = True
UNASSIGNED_LABEL_CANDIDATES = ["Unassigned", "Uncategorised", "Uncategorized"]
UNASSIGNED_GATE_MODE        = "hard"   # "hard" or "relative"
UNASSIGNED_MARGIN           = 0.05
EVAL_THRESHOLDS             = None     # will be filled after calibration

# =========================
# Path helpers
# =========================
def resolve_regex_dir():
    cands = []
    if os.path.isdir(PREFERRED_REGEX_DIR):
        cands.append(PREFERRED_REGEX_DIR)
    if not cands:
        for mp in glob.glob("**/regex_split_meta.json", recursive=True):
            cands.append(os.path.dirname(mp))
    for d in cands:
        req = [os.path.join(d, "regex_split_meta.json"),
               os.path.join(d, "train_BERTJE_REGEX_CHUNKS.csv"),
               os.path.join(d, "val_BERTJE_REGEX_CHUNKS.csv")]
        if all(os.path.exists(p) for p in req):
            return d
    raise FileNotFoundError("Could not find REGEX data folder with required files.")

def ensure_checkpoint_base(base_dir, model_name=SBERT_MODEL_NAME):
    os.makedirs(base_dir, exist_ok=True)
    if not any(os.scandir(base_dir)):
        print(f"[init] Seeding empty checkpoint base with tokenizer+config from {model_name}")
        tok = AutoTokenizer.from_pretrained(model_name)
        cfg = AutoConfig.from_pretrained(model_name)
        tok.save_pretrained(base_dir)
        cfg.save_pretrained(base_dir)
    return base_dir

def resolve_checkpoint_dir(base_dir):
    os.makedirs(base_dir, exist_ok=True)
    cks = sorted(glob.glob(os.path.join(base_dir, "checkpoint-*")), key=os.path.getmtime, reverse=True)
    if cks:
        print("Using latest checkpoint:", cks[0])
        return cks[0]
    if os.path.exists(os.path.join(base_dir, "config.json")):
        print("Using base dir (tokenizer+config):", base_dir)
        return base_dir
    return ensure_checkpoint_base(base_dir)

def pick_paths(labeling_system: str, regex_dir: str):
    if labeling_system == "cosine":
        return COSINE_TRAIN_PATH, COSINE_VAL_PATH, COSINE_META_PATH, "cosine"
    if labeling_system == "regex":
        return (os.path.join(regex_dir, "train_BERTJE_REGEX_CHUNKS.csv"),
                os.path.join(regex_dir, "val_BERTJE_REGEX_CHUNKS.csv"),
                os.path.join(regex_dir, "regex_split_meta.json"),
                "regex")
    try:
        d = resolve_regex_dir()
        return (os.path.join(d, "train_BERTJE_REGEX_CHUNKS.csv"),
                os.path.join(d, "val_BERTJE_REGEX_CHUNKS.csv"),
                os.path.join(d, "regex_split_meta.json"),
                "regex")
    except Exception:
        return COSINE_TRAIN_PATH, COSINE_VAL_PATH, COSINE_META_PATH, "cosine"

# =========================
# Safe parsers
# =========================
def safe_parse_list(x):
    if isinstance(x, (list, tuple, np.ndarray)):
        return [float(i) for i in x]
    if isinstance(x, str):
        s = x.strip()
        try:
            v = ast.literal_eval(s)
            if isinstance(v, (list, tuple, np.ndarray)):
                return [float(i) for i in v]
        except Exception:
            pass
        parts = [p for p in s.replace("[","").replace("]","").replace("(","").replace(")","").split(",") if p.strip()]
        try:
            return [float(p.strip()) for p in parts]
        except Exception:
            return []
    return []

def ensure_label_columns(df, use_multi: bool):
    if use_multi:
        if "labels" not in df.columns:
            raise ValueError("Multi-label mode requires a 'labels' column.")
        df = df.copy()
        df["labels"] = df["labels"].apply(lambda z: [int(round(v)) for v in safe_parse_list(z)])
    else:
        if "label" not in df.columns:
            raise ValueError("Single-label mode requires a 'label' column.")
        df = df.dropna(subset=["label"]).copy()
        df["label"] = df["label"].astype(int)
    return df

def load_meta(meta_path):
    with open(meta_path, "r", encoding="utf-8") as f:
        meta = json.load(f)
    topic_names   = meta["topic_names"]
    class_weights = meta.get("class_weights", [1.0]*len(topic_names))
    use_multi     = meta.get("use_multi_label", True)
    none_label    = meta.get("none_label_name") or meta.get("uncategorised_topic_name") or None
    return topic_names, torch.tensor(class_weights, dtype=torch.float), bool(use_multi), meta, none_label

# =========================
# Resolve paths + load meta/data
# =========================
CHECKPOINT_BASE = ensure_checkpoint_base(CHECKPOINT_BASE, SBERT_MODEL_NAME)
REGEX_DIR = resolve_regex_dir()
TRAIN_PATH, VAL_PATH, META_PATH, SYSTEM = pick_paths(LABELING_SYSTEM, REGEX_DIR)
CHECKPOINT_PATH = resolve_checkpoint_dir(CHECKPOINT_BASE)

print(f"\n=== RESOLVED PATHS ===")
print(f"LABELING_SYSTEM: {SYSTEM}")
print(f"REGEX_DIR:       {REGEX_DIR}")
print(f"TRAIN_PATH:      {TRAIN_PATH}")
print(f"VAL_PATH:        {VAL_PATH}")
print(f"META_PATH:       {META_PATH}")
print(f"CHECKPOINT_BASE: {CHECKPOINT_BASE}")
print(f"MODEL_LOAD_PATH: {CHECKPOINT_PATH}\n")

topic_names, class_weights, meta_multi, meta_dict, meta_none_name = load_meta(META_PATH)
num_labels = len(topic_names)

UNASSIGNED_IDX = None
cand_names = []
if meta_none_name:
    cand_names.append(meta_none_name)
cand_names.extend([n for n in UNASSIGNED_LABEL_CANDIDATES if n not in cand_names])
for nm in cand_names:
    if nm in topic_names:
        UNASSIGNED_IDX = topic_names.index(nm)
        break
print(f"[INFO] Unassigned label index: {UNASSIGNED_IDX} ({topic_names[UNASSIGNED_IDX] if UNASSIGNED_IDX is not None else 'N/A'})")

if USE_MULTI_LABEL == "auto":
    USE_MULTI_LABEL = meta_multi
else:
    USE_MULTI_LABEL = bool(str(USE_MULTI_LABEL).lower() == "true")
print(f"Loaded {num_labels} topics. Multi-label: {USE_MULTI_LABEL}")

train_df = pd.read_csv(TRAIN_PATH)
val_df   = pd.read_csv(VAL_PATH)
if "text_for_match" not in train_df.columns or "text_for_match" not in val_df.columns:
    raise ValueError("Both train and val CSVs must contain a 'text_for_match' column.")
train_df = ensure_label_columns(train_df, USE_MULTI_LABEL)
val_df   = ensure_label_columns(val_df,   USE_MULTI_LABEL)

# =========================
# Tokenizer & datasets
# =========================
tokenizer = AutoTokenizer.from_pretrained(SBERT_MODEL_NAME)

def _coerce_text_column(df: pd.DataFrame) -> pd.DataFrame:
    col = "text_for_match"
    if col not in df.columns:
        raise ValueError(f"Missing '{col}' column.")
    def to_str(x):
        if x is None:
            return ""
        try:
            if pd.isna(x):
                return ""
        except Exception:
            pass
        return x if isinstance(x, str) else str(x)
    out = df.copy()
    out[col] = out[col].apply(to_str)
    return out

train_df = _coerce_text_column(train_df)
val_df   = _coerce_text_column(val_df)

# 2) REPLACE your tokenize_function with truncation-only (no padding/max_length)
def tokenize_function(batch):
    return tokenizer(batch["text_for_match"], truncation=True)


train_ds = Dataset.from_pandas(train_df, preserve_index=False)
val_ds   = Dataset.from_pandas(val_df,   preserve_index=False)

keep_cols = ["text_for_match", "labels", "label"]
train_remove = [c for c in train_ds.column_names if c not in keep_cols]
val_remove   = [c for c in val_ds.column_names if c not in keep_cols]

train_ds = train_ds.map(tokenize_function, batched=True, remove_columns=train_remove)
val_ds   = val_ds.map(tokenize_function,   batched=True, remove_columns=val_remove)

cols_to_keep = ["input_ids", "attention_mask", "labels"] if USE_MULTI_LABEL else ["input_ids", "attention_mask", "label"]
train_ds = train_ds.remove_columns([c for c in train_ds.column_names if c not in cols_to_keep])
val_ds   = val_ds.remove_columns([c for c in val_ds.column_names   if c not in cols_to_keep])

# =========================
# SBERT backbone + pooling + multilabel head
# =========================
class MeanPooling(nn.Module):
    def forward(self, last_hidden_state, attention_mask):
        mask = attention_mask.unsqueeze(-1).type_as(last_hidden_state)  # [B,S,1]
        summed = (last_hidden_state * mask).sum(dim=1)                  # [B,H]
        counts = mask.sum(dim=1).clamp(min=1e-9)                        # [B,1]
        return summed / counts

class SBERTMultiLabelClassifier(nn.Module):
    def __init__(self, base_name: str, num_labels: int, dropout: float = 0.1):
        super().__init__()
        self.encoder = AutoModel.from_pretrained(base_name)
        hidden = self.encoder.config.hidden_size
        self.pool = MeanPooling()
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(hidden, num_labels)

        self.config = AutoConfig.from_pretrained(base_name)
        self.config.num_labels = num_labels
        self.config.problem_type = "multi_label_classification"  # for completeness

        self.id2label = None
        self.label2id = None

    def forward(self, input_ids=None, attention_mask=None, **kwargs):
        out = self.encoder(input_ids=input_ids, attention_mask=attention_mask, return_dict=True)
        sent = self.pool(out.last_hidden_state, attention_mask)
        sent = self.dropout(sent)
        logits = self.classifier(sent)
        return SequenceClassifierOutput(logits=logits)

model = SBERTMultiLabelClassifier(SBERT_MODEL_NAME, num_labels=num_labels)

# Try to resume safely from checkpoint (if pytorch_model.bin exists)
def try_load_state_dict_if_available(model, checkpoint_dir):
    candidate_dirs = []
    if os.path.isdir(checkpoint_dir):
        sub = sorted(glob.glob(os.path.join(checkpoint_dir, "checkpoint-*")), key=os.path.getmtime, reverse=True)
        candidate_dirs.extend(sub)
        candidate_dirs.append(checkpoint_dir)
    for d in candidate_dirs:
        bin_path = os.path.join(d, "pytorch_model.bin")
        if os.path.exists(bin_path):
            try:
                print(f"[resume] Loading state_dict from {bin_path}")
                state = torch.load(bin_path, map_location="cpu")
                missing, unexpected = model.load_state_dict(state, strict=False)
                print(f"[resume] Loaded with missing={len(missing)}, unexpected={len(unexpected)}")
                return True
            except Exception as e:
                print(f"[resume] Failed to load from {bin_path}: {e}")
    return False

_ = try_load_state_dict_if_available(model, CHECKPOINT_PATH)

# id2label / label2id (helpful metadata)
id2label = {i: n for i, n in enumerate(topic_names)}
label2id = {n: i for i, n in enumerate(topic_names)}
model.id2label = id2label
model.label2id = label2id

# =========================
# Metrics (+ Unassigned gate)
# =========================
def apply_unassigned_gate(probs: np.ndarray, preds: np.ndarray, thresholds: np.ndarray) -> np.ndarray:
    if not (USE_UNASSIGNED_GATE and UNASSIGNED_IDX is not None):
        return preds
    pu = probs[:, UNASSIGNED_IDX]
    tu = thresholds[UNASSIGNED_IDX]
    if UNASSIGNED_GATE_MODE == "hard":
        gate = pu >= tu
    else:
        mask = np.ones(probs.shape[1], dtype=bool)
        mask[UNASSIGNED_IDX] = False
        max_other = probs[:, mask].max(axis=1) if mask.any() else 0.0
        gate = pu >= (max_other + UNASSIGNED_MARGIN)
    preds[gate, :] = 0
    preds[gate, UNASSIGNED_IDX] = 1
    return preds

# =========================
# Metrics (+ Unassigned gate)
# =========================
def apply_unassigned_gate(probs: np.ndarray, preds: np.ndarray, thresholds: np.ndarray) -> np.ndarray:
    if not (USE_UNASSIGNED_GATE and UNASSIGNED_IDX is not None):
        return preds
    pu = probs[:, UNASSIGNED_IDX]
    tu = thresholds[UNASSIGNED_IDX]
    if UNASSIGNED_GATE_MODE == "hard":
        gate = pu >= tu
    else:
        mask = np.ones(probs.shape[1], dtype=bool)
        mask[UNASSIGNED_IDX] = False
        max_other = probs[:, mask].max(axis=1) if mask.any() else 0.0
        gate = pu >= (max_other + UNASSIGNED_MARGIN)
    preds[gate, :] = 0
    preds[gate, UNASSIGNED_IDX] = 1
    return preds

def compute_metrics(eval_pred):
    # robustly accept both EvalPrediction and tuple
    if isinstance(eval_pred, EvalPrediction):
        logits, labels = eval_pred.predictions, eval_pred.label_ids
    else:
        logits, labels = eval_pred

    import numpy as np
    from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score
    import torch as _torch

    # Compute loss for metric tracking (REQUIRED for metric_for_best_model)
    logits_tensor = _torch.tensor(logits, dtype=_torch.float32)
    
    metrics = {}
    if USE_MULTI_LABEL:
        labels_tensor = _torch.tensor(labels, dtype=_torch.float32)
        loss_fn = nn.BCEWithLogitsLoss(pos_weight=class_weights)
        eval_loss = loss_fn(logits_tensor, labels_tensor).item()
        
        probs = _torch.sigmoid(logits_tensor).numpy()
        thr   = np.array(EVAL_THRESHOLDS, dtype=float) if EVAL_THRESHOLDS is not None else np.full((num_labels,), 0.5, dtype=float)
        preds = (probs >= thr[None, :]).astype(int)
        preds = apply_unassigned_gate(probs, preds, thr)
        labels_int = labels.astype(int)
        
        metrics = {
            "eval_loss": eval_loss,  # CRITICAL: must include this for checkpoint selection
            "accuracy":  accuracy_score(labels_int, preds),
            "precision": precision_score(labels_int, preds, average="weighted", zero_division=0),
            "recall":    recall_score(labels_int, preds, average="weighted", zero_division=0),
            "f1":        f1_score(labels_int, preds, average="weighted", zero_division=0),
        }
    else:
        labels_tensor = _torch.tensor(labels, dtype=_torch.long)
        loss_fn = nn.CrossEntropyLoss(weight=class_weights)
        eval_loss = loss_fn(logits_tensor, labels_tensor).item()
        
        preds = np.argmax(logits, axis=1)
        metrics = {
            "eval_loss": eval_loss,  # CRITICAL: must include this for checkpoint selection
            "accuracy":  accuracy_score(labels, preds),
            "precision": precision_score(labels, preds, average="weighted", zero_division=0),
            "recall":    recall_score(labels, preds, average="weighted", zero_division=0),
            "f1":        f1_score(labels, preds, average="weighted", zero_division=0),
        }

    return metrics


# 4) CREATE a data collator that pads each batch to that batch’s longest sequence
# (pad_to_multiple_of=8 is optional; it can speed up fp16/bf16 on GPUs)
data_collator = DataCollatorWithPadding(tokenizer, pad_to_multiple_of=8)


# =========================
# Custom Trainer with weighted loss
# =========================
class CustomLossTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        model_inputs = {k: v for k, v in inputs.items() if k in ["input_ids","attention_mask"]}
        outputs = model(**model_inputs)
        logits = outputs.logits

        if USE_MULTI_LABEL:
            labels = inputs["labels"]
            labels = labels.float() if hasattr(labels, "float") else torch.tensor(labels, dtype=torch.float)
            labels = labels.to(logits.device)
            loss_fn = nn.BCEWithLogitsLoss(pos_weight=class_weights.to(logits.device))
            loss = loss_fn(logits, labels)
        else:
            labels = inputs["label"]
            labels = labels.long() if hasattr(labels, "long") else torch.tensor(labels, dtype=torch.long)
            labels = labels.to(logits.device)
            loss_fn = nn.CrossEntropyLoss(weight=class_weights.to(logits.device))
            loss = loss_fn(logits, labels)

        # CRITICAL: Must return outputs with logits for compute_metrics to work
        if return_outputs:
            # Ensure outputs has the logits attribute accessible
            return loss, {"logits": logits}
        return loss
    
    def prediction_step(self, model, inputs, prediction_loss_only, ignore_keys=None):
        """
        Override prediction_step to ensure labels are properly handled
        """
        has_labels = "labels" in inputs if USE_MULTI_LABEL else "label" in inputs
        
        # Get the labels before removing them
        if USE_MULTI_LABEL:
            labels = inputs.get("labels")
        else:
            labels = inputs.get("label")
        
        # Forward pass without labels
        model_inputs = {k: v for k, v in inputs.items() if k in ["input_ids", "attention_mask"]}
        
        with torch.no_grad():
            outputs = model(**model_inputs)
            logits = outputs.logits
            
            # Compute loss manually
            if has_labels:
                if USE_MULTI_LABEL:
                    labels_tensor = labels.float().to(logits.device)
                    loss_fn = nn.BCEWithLogitsLoss(pos_weight=class_weights.to(logits.device))
                    loss = loss_fn(logits, labels_tensor)
                else:
                    labels_tensor = labels.long().to(logits.device)
                    loss_fn = nn.CrossEntropyLoss(weight=class_weights.to(logits.device))
                    loss = loss_fn(logits, labels_tensor)
            else:
                loss = None
        
        if prediction_loss_only:
            return (loss, None, None)
        
        # Return loss, logits, and labels for compute_metrics
        return (loss, logits.detach(), labels.detach() if labels is not None else None)

# =========================
# Training (safe: select best by eval_loss)
# =========================
os.environ["WANDB_DISABLED"] = "true"
training_args = TrainingArguments(
    output_dir= CHECKPOINT_SAVE_TO,      # write new checkpoint-* here
    evaluation_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=8,
    weight_decay=0.01,
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    logging_steps=50,
    save_total_limit=6,
    remove_unused_columns=False, 
)

trainer = CustomLossTrainer(
    model=model,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=val_ds,
    compute_metrics=compute_metrics,
    data_collator=data_collator,
)

# Train the model and ensure eval_loss is computed by using Trainer's internal evaluation
trainer.train()
trainer.remove_callback(PrinterCallback)

print("\n[EVAL @ end of training]")
print(trainer.evaluate())  # should include eval_loss and eval_f1, etc.

# =========================
# Post-training calibration + reports
# =========================
eval_results = trainer.predict(val_ds)
logits = torch.tensor(eval_results.predictions)

from sklearn.metrics import f1_score, confusion_matrix
if USE_MULTI_LABEL:
    probs  = torch.sigmoid(logits).numpy()
    y_true = eval_results.label_ids.astype(int)

    def find_best_thresholds(y_true, probs, grid=np.linspace(0.05, 0.95, 19)):
        C = y_true.shape[1]
        best = [0.5]*C
        for c in range(C):
            y = y_true[:, c]
            if y.sum() == 0:
                continue
            p = probs[:, c]
            best_f1, best_t = 0.0, 0.5
            for t in grid:
                f1 = f1_score(y, (p >= t).astype(int), zero_division=0)
                if f1 > best_f1:
                    best_f1, best_t = f1, t
            best[c] = float(best_t)
        return best

    thresholds = find_best_thresholds(y_true, probs)
    thr_arr = np.array(thresholds, dtype=float)
    y_pred = (probs >= thr_arr[None, :]).astype(int)
    y_pred = apply_unassigned_gate(probs, y_pred, thr_arr)

    print("\nPer-class thresholds (incl. Unassigned if present):", thresholds)
    print(classification_report(y_true, y_pred, target_names=topic_names, zero_division=0))

    EVAL_THRESHOLDS = thresholds
    os.makedirs(CHECKPOINT_BASE, exist_ok=True)
    with open(os.path.join(CHECKPOINT_BASE, "thresholds.json"), "w", encoding="utf-8") as f:
        json.dump({
            "topic_names": topic_names,
            "thresholds": thresholds,
            "use_unassigned_gate": USE_UNASSIGNED_GATE,
            "unassigned_idx": UNASSIGNED_IDX,
            "unassigned_gate_mode": UNASSIGNED_GATE_MODE,
            "unassigned_margin": UNASSIGNED_MARGIN
        }, f, indent=2, ensure_ascii=False)

    # Confusion matrix per class (binary)
    for idx, name in enumerate(topic_names):
        cm = confusion_matrix(y_true[:, idx], y_pred[:, idx])
        plt.figure(figsize=(4,3))
        sns.heatmap(cm, annot=True, fmt='d', cmap="Blues")
        plt.title(f"Confusion Matrix: {name}")
        plt.tight_layout()
        plt.show()
else:
    y_true = eval_results.label_ids
    y_pred = np.argmax(logits.numpy(), axis=1)
    print(classification_report(y_true, y_pred, target_names=topic_names, zero_division=0))
    cm = confusion_matrix(y_true, y_pred, labels=range(len(topic_names)))
    plt.figure(figsize=(12,10))
    sns.heatmap(cm, annot=True, fmt='d', xticklabels=topic_names, yticklabels=topic_names, cmap="Blues")
    plt.title("Confusion Matrix: Topic Classification")
    plt.tight_layout()
    plt.show()

print("Done.")

# =========================
# (Optional) If you now see 'eval_f1' in trainer.evaluate(), you can switch to F1 selection:
# Rebuild TrainingArguments with:
#   metric_for_best_model="f1", greater_is_better=True
# and recreate the Trainer before training again.
# =========================
```

```python
# Train SBERT-style classifier on EITHER cosine-labeled OR regex-labeled datasets (toggle via CONFIG)
# Robust path resolution + checkpoint auto-creation (tokenizer+config) + safe loading
# Loss entirely in CustomLossTrainer. Includes "Unassigned" gate + per-class threshold calibration.
# Safe best-model selection via eval_loss (guaranteed), with optional switch to F1 after verifying metrics.

import os, json, ast, glob
import numpy as np
import pandas as pd
import torch
from torch import nn
from datasets import Dataset
from transformers import (
    AutoModel,
    AutoTokenizer,
    AutoConfig,
    TrainingArguments,
    Trainer,
    EvalPrediction
)
from transformers.modeling_outputs import SequenceClassifierOutput
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt
import seaborn as sns
# 1) ADD this import (near your other HF imports)
from transformers import DataCollatorWithPadding


# =========================
# CONFIG
# =========================
LABELING_SYSTEM = "regex"    # "cosine", "regex", or "auto"
USE_MULTI_LABEL = "true"     # "auto" | True | False

# SBERT backbone (Dutch-capable encoders recommended)
SBERT_MODEL_NAME = "GroNLP/bert-base-dutch-cased"
# Alternatives: "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2", etc.

CHECKPOINT_BASE    = r"C:\Users\Home\policy-analysis\Modeling\Regex_pretraining_slavery_sentences_v4\bertje_checkpoints"
CHECKPOINT_SAVE_TO = r"C:\Users\Home\policy-analysis\Modeling\Regex_pretraining_slavery_sentences_v4\bertje_checkpoints"

# Cosine defaults
COSINE_DIR        = r"C:\Users\Home\policy-analysis\Corex_topicdata_slavery_v5"
COSINE_TRAIN_PATH = os.path.join(COSINE_DIR, "train_BERTJE_CHUNKS.csv")
COSINE_VAL_PATH   = os.path.join(COSINE_DIR, "val_BERTJE_CHUNKS.csv")
COSINE_META_PATH  = r"C:\Users\Home\policy-analysis\Corex_topicdata_Slavery\bertje_split_meta.json"

# Regex preferred dir
PREFERRED_REGEX_DIR = r"C:\Users\Home\policy-analysis\Modeling\Regex_pretraining_slavery_sentences_v4\Bertje_prep"

# "Unassigned" (NONE-OF-THE-ABOVE) gate
USE_UNASSIGNED_GATE         = True
UNASSIGNED_LABEL_CANDIDATES = ["Unassigned", "Uncategorised", "Uncategorized"]
UNASSIGNED_GATE_MODE        = "hard"   # "hard" or "relative"
UNASSIGNED_MARGIN           = 0.05
EVAL_THRESHOLDS             = None     # will be filled after calibration

# =========================
# Path helpers
# =========================
def resolve_regex_dir():
    cands = []
    if os.path.isdir(PREFERRED_REGEX_DIR):
        cands.append(PREFERRED_REGEX_DIR)
    if not cands:
        for mp in glob.glob("**/regex_split_meta.json", recursive=True):
            cands.append(os.path.dirname(mp))
    for d in cands:
        req = [os.path.join(d, "regex_split_meta.json"),
               os.path.join(d, "train_BERTJE_REGEX_CHUNKS.csv"),
               os.path.join(d, "val_BERTJE_REGEX_CHUNKS.csv")]
        if all(os.path.exists(p) for p in req):
            return d
    raise FileNotFoundError("Could not find REGEX data folder with required files.")

def ensure_checkpoint_base(base_dir, model_name=SBERT_MODEL_NAME):
    os.makedirs(base_dir, exist_ok=True)
    if not any(os.scandir(base_dir)):
        print(f"[init] Seeding empty checkpoint base with tokenizer+config from {model_name}")
        tok = AutoTokenizer.from_pretrained(model_name)
        cfg = AutoConfig.from_pretrained(model_name)
        tok.save_pretrained(base_dir)
        cfg.save_pretrained(base_dir)
    return base_dir

def resolve_checkpoint_dir(base_dir):
    os.makedirs(base_dir, exist_ok=True)
    cks = sorted(glob.glob(os.path.join(base_dir, "checkpoint-*")), key=os.path.getmtime, reverse=True)
    if cks:
        print("Using latest checkpoint:", cks[0])
        return cks[0]
    if os.path.exists(os.path.join(base_dir, "config.json")):
        print("Using base dir (tokenizer+config):", base_dir)
        return base_dir
    return ensure_checkpoint_base(base_dir)

def pick_paths(labeling_system: str, regex_dir: str):
    if labeling_system == "cosine":
        return COSINE_TRAIN_PATH, COSINE_VAL_PATH, COSINE_META_PATH, "cosine"
    if labeling_system == "regex":
        return (os.path.join(regex_dir, "train_BERTJE_REGEX_CHUNKS.csv"),
                os.path.join(regex_dir, "val_BERTJE_REGEX_CHUNKS.csv"),
                os.path.join(regex_dir, "regex_split_meta.json"),
                "regex")
    try:
        d = resolve_regex_dir()
        return (os.path.join(d, "train_BERTJE_REGEX_CHUNKS.csv"),
                os.path.join(d, "val_BERTJE_REGEX_CHUNKS.csv"),
                os.path.join(d, "regex_split_meta.json"),
                "regex")
    except Exception:
        return COSINE_TRAIN_PATH, COSINE_VAL_PATH, COSINE_META_PATH, "cosine"

# =========================
# Safe parsers
# =========================
def safe_parse_list(x):
    if isinstance(x, (list, tuple, np.ndarray)):
        return [float(i) for i in x]
    if isinstance(x, str):
        s = x.strip()
        try:
            v = ast.literal_eval(s)
            if isinstance(v, (list, tuple, np.ndarray)):
                return [float(i) for i in v]
        except Exception:
            pass
        parts = [p for p in s.replace("[","").replace("]","").replace("(","").replace(")","").split(",") if p.strip()]
        try:
            return [float(p.strip()) for p in parts]
        except Exception:
            return []
    return []

def ensure_label_columns(df, use_multi: bool):
    if use_multi:
        if "labels" not in df.columns:
            raise ValueError("Multi-label mode requires a 'labels' column.")
        df = df.copy()
        df["labels"] = df["labels"].apply(lambda z: [int(round(v)) for v in safe_parse_list(z)])
    else:
        if "label" not in df.columns:
            raise ValueError("Single-label mode requires a 'label' column.")
        df = df.dropna(subset=["label"]).copy()
        df["label"] = df["label"].astype(int)
    return df

def load_meta(meta_path):
    with open(meta_path, "r", encoding="utf-8") as f:
        meta = json.load(f)
    topic_names   = meta["topic_names"]
    class_weights = meta.get("class_weights", [1.0]*len(topic_names))
    use_multi     = meta.get("use_multi_label", True)
    none_label    = meta.get("none_label_name") or meta.get("uncategorised_topic_name") or None
    return topic_names, torch.tensor(class_weights, dtype=torch.float), bool(use_multi), meta, none_label

# =========================
# Resolve paths + load meta/data
# =========================
CHECKPOINT_BASE = ensure_checkpoint_base(CHECKPOINT_BASE, SBERT_MODEL_NAME)
REGEX_DIR = resolve_regex_dir()
TRAIN_PATH, VAL_PATH, META_PATH, SYSTEM = pick_paths(LABELING_SYSTEM, REGEX_DIR)
CHECKPOINT_PATH = resolve_checkpoint_dir(CHECKPOINT_BASE)

print(f"\n=== RESOLVED PATHS ===")
print(f"LABELING_SYSTEM: {SYSTEM}")
print(f"REGEX_DIR:       {REGEX_DIR}")
print(f"TRAIN_PATH:      {TRAIN_PATH}")
print(f"VAL_PATH:        {VAL_PATH}")
print(f"META_PATH:       {META_PATH}")
print(f"CHECKPOINT_BASE: {CHECKPOINT_BASE}")
print(f"MODEL_LOAD_PATH: {CHECKPOINT_PATH}\n")

topic_names, class_weights, meta_multi, meta_dict, meta_none_name = load_meta(META_PATH)
num_labels = len(topic_names)

UNASSIGNED_IDX = None
cand_names = []
if meta_none_name:
    cand_names.append(meta_none_name)
cand_names.extend([n for n in UNASSIGNED_LABEL_CANDIDATES if n not in cand_names])
for nm in cand_names:
    if nm in topic_names:
        UNASSIGNED_IDX = topic_names.index(nm)
        break
print(f"[INFO] Unassigned label index: {UNASSIGNED_IDX} ({topic_names[UNASSIGNED_IDX] if UNASSIGNED_IDX is not None else 'N/A'})")

if USE_MULTI_LABEL == "auto":
    USE_MULTI_LABEL = meta_multi
else:
    USE_MULTI_LABEL = bool(str(USE_MULTI_LABEL).lower() == "true")
print(f"Loaded {num_labels} topics. Multi-label: {USE_MULTI_LABEL}")

train_df = pd.read_csv(TRAIN_PATH)
val_df   = pd.read_csv(VAL_PATH)
if "text_for_match" not in train_df.columns or "text_for_match" not in val_df.columns:
    raise ValueError("Both train and val CSVs must contain a 'text_for_match' column.")
train_df = ensure_label_columns(train_df, USE_MULTI_LABEL)
val_df   = ensure_label_columns(val_df,   USE_MULTI_LABEL)

# =========================
# Tokenizer & datasets
# =========================
tokenizer = AutoTokenizer.from_pretrained(SBERT_MODEL_NAME)

def _coerce_text_column(df: pd.DataFrame) -> pd.DataFrame:
    col = "text_for_match"
    if col not in df.columns:
        raise ValueError(f"Missing '{col}' column.")
    def to_str(x):
        if x is None:
            return ""
        try:
            if pd.isna(x):
                return ""
        except Exception:
            pass
        return x if isinstance(x, str) else str(x)
    out = df.copy()
    out[col] = out[col].apply(to_str)
    return out

train_df = _coerce_text_column(train_df)
val_df   = _coerce_text_column(val_df)

# 2) REPLACE your tokenize_function with truncation-only (no padding/max_length)
def tokenize_function(batch):
    return tokenizer(batch["text_for_match"], truncation=True)


train_ds = Dataset.from_pandas(train_df, preserve_index=False)
val_ds   = Dataset.from_pandas(val_df,   preserve_index=False)

keep_cols = ["text_for_match", "labels", "label"]
train_remove = [c for c in train_ds.column_names if c not in keep_cols]
val_remove   = [c for c in val_ds.column_names if c not in keep_cols]

train_ds = train_ds.map(tokenize_function, batched=True, remove_columns=train_remove)
val_ds   = val_ds.map(tokenize_function,   batched=True, remove_columns=val_remove)

cols_to_keep = ["input_ids", "attention_mask", "labels"] if USE_MULTI_LABEL else ["input_ids", "attention_mask", "label"]
train_ds = train_ds.remove_columns([c for c in train_ds.column_names if c not in cols_to_keep])
val_ds   = val_ds.remove_columns([c for c in val_ds.column_names   if c not in cols_to_keep])

# =========================
# SBERT backbone + pooling + multilabel head
# =========================
class MeanPooling(nn.Module):
    def forward(self, last_hidden_state, attention_mask):
        mask = attention_mask.unsqueeze(-1).type_as(last_hidden_state)  # [B,S,1]
        summed = (last_hidden_state * mask).sum(dim=1)                  # [B,H]
        counts = mask.sum(dim=1).clamp(min=1e-9)                        # [B,1]
        return summed / counts

class SBERTMultiLabelClassifier(nn.Module):
    def __init__(self, base_name: str, num_labels: int, dropout: float = 0.1):
        super().__init__()
        self.encoder = AutoModel.from_pretrained(base_name)
        hidden = self.encoder.config.hidden_size
        self.pool = MeanPooling()
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(hidden, num_labels)

        self.config = AutoConfig.from_pretrained(base_name)
        self.config.num_labels = num_labels
        self.config.problem_type = "multi_label_classification"  # for completeness

        self.id2label = None
        self.label2id = None

    def forward(self, input_ids=None, attention_mask=None, **kwargs):
        out = self.encoder(input_ids=input_ids, attention_mask=attention_mask, return_dict=True)
        sent = self.pool(out.last_hidden_state, attention_mask)
        sent = self.dropout(sent)
        logits = self.classifier(sent)
        return SequenceClassifierOutput(logits=logits)

model = SBERTMultiLabelClassifier(SBERT_MODEL_NAME, num_labels=num_labels)

# Try to resume safely from checkpoint (if pytorch_model.bin exists)
def try_load_state_dict_if_available(model, checkpoint_dir):
    candidate_dirs = []
    if os.path.isdir(checkpoint_dir):
        sub = sorted(glob.glob(os.path.join(checkpoint_dir, "checkpoint-*")), key=os.path.getmtime, reverse=True)
        candidate_dirs.extend(sub)
        candidate_dirs.append(checkpoint_dir)
    for d in candidate_dirs:
        bin_path = os.path.join(d, "pytorch_model.bin")
        if os.path.exists(bin_path):
            try:
                print(f"[resume] Loading state_dict from {bin_path}")
                state = torch.load(bin_path, map_location="cpu")
                missing, unexpected = model.load_state_dict(state, strict=False)
                print(f"[resume] Loaded with missing={len(missing)}, unexpected={len(unexpected)}")
                return True
            except Exception as e:
                print(f"[resume] Failed to load from {bin_path}: {e}")
    return False

_ = try_load_state_dict_if_available(model, CHECKPOINT_PATH)

# id2label / label2id (helpful metadata)
id2label = {i: n for i, n in enumerate(topic_names)}
label2id = {n: i for i, n in enumerate(topic_names)}
model.id2label = id2label
model.label2id = label2id

# =========================
# Metrics (+ Unassigned gate)
# =========================
def apply_unassigned_gate(probs: np.ndarray, preds: np.ndarray, thresholds: np.ndarray) -> np.ndarray:
    if not (USE_UNASSIGNED_GATE and UNASSIGNED_IDX is not None):
        return preds
    pu = probs[:, UNASSIGNED_IDX]
    tu = thresholds[UNASSIGNED_IDX]
    if UNASSIGNED_GATE_MODE == "hard":
        gate = pu >= tu
    else:
        mask = np.ones(probs.shape[1], dtype=bool)
        mask[UNASSIGNED_IDX] = False
        max_other = probs[:, mask].max(axis=1) if mask.any() else 0.0
        gate = pu >= (max_other + UNASSIGNED_MARGIN)
    preds[gate, :] = 0
    preds[gate, UNASSIGNED_IDX] = 1
    return preds

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    if USE_MULTI_LABEL:
        probs = torch.sigmoid(torch.tensor(logits)).numpy()
        # thresholds during training/eval step
        thr = np.array(EVAL_THRESHOLDS, dtype=float) if EVAL_THRESHOLDS is not None else np.full((num_labels,), 0.5, dtype=float)
        preds = (probs >= thr[None, :]).astype(int)
        preds = apply_unassigned_gate(probs, preds, thr)
        labels = labels.astype(int)
        from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score
        return {
            "accuracy": accuracy_score(labels, preds),
            "precision": precision_score(labels, preds, average="weighted", zero_division=0),
            "recall": recall_score(labels, preds, average="weighted", zero_division=0),
            "f1": f1_score(labels, preds, average="weighted", zero_division=0),
        }
    else:
        preds = np.argmax(logits, axis=1)
        from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score
        return {
            "accuracy": accuracy_score(labels, preds),
            "precision": precision_score(labels, preds, average="weighted", zero_division=0),
            "recall": recall_score(labels, preds, average="weighted", zero_division=0),
            "f1": f1_score(labels, preds, average="weighted")
        }



# 4) CREATE a data collator that pads each batch to that batch’s longest sequence
# (pad_to_multiple_of=8 is optional; it can speed up fp16/bf16 on GPUs)
data_collator = DataCollatorWithPadding(tokenizer, pad_to_multiple_of=8)


# =========================
# Custom Trainer with weighted loss
# =========================
class CustomLossTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        model_inputs = {k: v for k, v in inputs.items() if k in ["input_ids","attention_mask"]}
        outputs = model(**model_inputs)
        logits = outputs.logits

        if USE_MULTI_LABEL:
            labels = inputs["labels"]
            labels = labels.float() if hasattr(labels, "float") else torch.tensor(labels, dtype=torch.float)
            labels = labels.to(logits.device)
            loss_fn = nn.BCEWithLogitsLoss(pos_weight=class_weights.to(logits.device))
            loss = loss_fn(logits, labels)
        else:
            labels = inputs["label"]
            labels = labels.long() if hasattr(labels, "long") else torch.tensor(labels, dtype=torch.long)
            labels = labels.to(logits.device)
            loss_fn = nn.CrossEntropyLoss(weight=class_weights.to(logits.device))
            loss = loss_fn(logits, labels)

        return (loss, outputs) if return_outputs else loss

# =========================
# Training (safe: select best by eval_loss)
# =========================
os.environ["WANDB_DISABLED"] = "true"
training_args = TrainingArguments(
    output_dir=CHECKPOINT_SAVE_TO,
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",  # guaranteed metric
    greater_is_better=False,
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_steps=50,
    save_total_limit=6,
    remove_unused_columns=False,        # keep 'labels' for the custom loss
)

trainer = CustomLossTrainer(
    model=model,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=val_ds,
    compute_metrics=compute_metrics,
    data_collator=data_collator,
)

trainer.train()
print("\n[EVAL @ end of training]")
print(trainer.evaluate())  # should include eval_loss and eval_f1, etc.

# =========================
# Post-training calibration + reports
# =========================
eval_results = trainer.predict(val_ds)
logits = torch.tensor(eval_results.predictions)

from sklearn.metrics import f1_score, confusion_matrix
if USE_MULTI_LABEL:
    probs  = torch.sigmoid(logits).numpy()
    y_true = eval_results.label_ids.astype(int)

    def find_best_thresholds(y_true, probs, grid=np.linspace(0.05, 0.95, 19)):
        C = y_true.shape[1]
        best = [0.5]*C
        for c in range(C):
            y = y_true[:, c]
            if y.sum() == 0:
                continue
            p = probs[:, c]
            best_f1, best_t = 0.0, 0.5
            for t in grid:
                f1 = f1_score(y, (p >= t).astype(int), zero_division=0)
                if f1 > best_f1:
                    best_f1, best_t = f1, t
            best[c] = float(best_t)
        return best

    thresholds = find_best_thresholds(y_true, probs)
    thr_arr = np.array(thresholds, dtype=float)
    y_pred = (probs >= thr_arr[None, :]).astype(int)
    y_pred = apply_unassigned_gate(probs, y_pred, thr_arr)

    print("\nPer-class thresholds (incl. Unassigned if present):", thresholds)
    print(classification_report(y_true, y_pred, target_names=topic_names, zero_division=0))

    EVAL_THRESHOLDS = thresholds
    os.makedirs(CHECKPOINT_BASE, exist_ok=True)
    with open(os.path.join(CHECKPOINT_BASE, "thresholds.json"), "w", encoding="utf-8") as f:
        json.dump({
            "topic_names": topic_names,
            "thresholds": thresholds,
            "use_unassigned_gate": USE_UNASSIGNED_GATE,
            "unassigned_idx": UNASSIGNED_IDX,
            "unassigned_gate_mode": UNASSIGNED_GATE_MODE,
            "unassigned_margin": UNASSIGNED_MARGIN
        }, f, indent=2, ensure_ascii=False)

    # Confusion matrix per class (binary)
    for idx, name in enumerate(topic_names):
        cm = confusion_matrix(y_true[:, idx], y_pred[:, idx])
        plt.figure(figsize=(4,3))
        sns.heatmap(cm, annot=True, fmt='d', cmap="Blues")
        plt.title(f"Confusion Matrix: {name}")
        plt.tight_layout()
        plt.show()
else:
    y_true = eval_results.label_ids
    y_pred = np.argmax(logits.numpy(), axis=1)
    print(classification_report(y_true, y_pred, target_names=topic_names, zero_division=0))
    cm = confusion_matrix(y_true, y_pred, labels=range(len(topic_names)))
    plt.figure(figsize=(12,10))
    sns.heatmap(cm, annot=True, fmt='d', xticklabels=topic_names, yticklabels=topic_names, cmap="Blues")
    plt.title("Confusion Matrix: Topic Classification")
    plt.tight_layout()
    plt.show()

print("Done.")

# =========================
# (Optional) If you now see 'eval_f1' in trainer.evaluate(), you can switch to F1 selection:
# Rebuild TrainingArguments with:
#   metric_for_best_model="f1", greater_is_better=True
# and recreate the Trainer before training again.
# =========================
```
